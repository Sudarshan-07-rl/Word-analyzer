```python
# ============================================================
# NLP MINI PROJECT
# WORD FREQUENCY ANALYZER
# ============================================================

# Step 1: Install NLTK
!pip install nltk

# Step 2: Import libraries
import re
import nltk
from collections import Counter
import matplotlib.pyplot as plt

# Step 3: Download stopwords
nltk.download('stopwords')

from nltk.corpus import stopwords


# Step 4: Take input text from user
text = input("Enter a paragraph: ")


# Step 5: Convert text into lowercase
text = text.lower()


# Step 6: Remove punctuation and numbers
text = re.sub(r'[^a-zA-Z\s]', '', text)


# Step 7: Split text into words
words = text.split()


# Step 8: Get English stop words
stop_words = set(stopwords.words('english'))


# Step 9: Remove stop words
filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)


# Step 10: Count frequency of each word
word_frequency = Counter(filtered_words)


# Step 11: Display total number of words
print("\nTotal Words:", len(words))


# Step 12: Display unique words
print("Unique Words:", len(set(words)))


# Step 13: Display word frequency
print("\nWord Frequency:")

for word, count in word_frequency.most_common():
    print(word, ":", count)


# Step 14: Get top 10 most frequent words
top_words = word_frequency.most_common(10)


# Step 15: Create lists for graph
words_for_graph = []
counts_for_graph = []

for word, count in top_words:
    words_for_graph.append(word)
    counts_for_graph.append(count)


# Step 16: Display graph
plt.figure(figsize=(10, 5))

plt.bar(words_for_graph, counts_for_graph)

plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 10 Most Frequent Words")

plt.xticks(rotation=45)

plt.show()
```

```python
# ============================================================
# NLP MINI PROJECT
# WORD FREQUENCY ANALYZER WITH GUI
# ============================================================

# STEP 1: Install libraries
!pip install -q gradio nltk

# STEP 2: Import libraries
import re
import nltk
import gradio as gr
import matplotlib.pyplot as plt

from collections import Counter
from nltk.corpus import stopwords

# STEP 3: Download stopwords
nltk.download('stopwords')

# Get English stop words
stop_words = set(stopwords.words('english'))


# STEP 4: Create analysis function
def analyze_text(text):

    # Check empty input
    if not text.strip():
        return "Please enter some text.", None

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Split text into words
    words = text.split()

    # Remove stop words
    filtered_words = [
        word for word in words
        if word not in stop_words
    ]

    # Count word frequency
    word_frequency = Counter(filtered_words)

    # Total words
    total_words = len(words)

    # Unique words
    unique_words = len(set(words))

    # Get top 10 words
    top_words = word_frequency.most_common(10)

    # Create result
    result = f"Total Words: {total_words}\n"
    result += f"Unique Words: {unique_words}\n\n"
    result += "Word Frequency:\n"

    for word, count in top_words:
        result += f"{word} : {count}\n"

    # Create graph
    word_list = [item[0] for item in top_words]
    count_list = [item[1] for item in top_words]

    plt.figure(figsize=(10, 5))
    plt.bar(word_list, count_list)

    plt.xlabel("Words")
    plt.ylabel("Frequency")
    plt.title("Top 10 Most Frequent Words")

    plt.xticks(rotation=45)
    plt.tight_layout()

    # Return result and graph
    return result, plt.gcf()


# STEP 5: Create GUI
interface = gr.Interface(
    fn=analyze_text,

    inputs=gr.Textbox(
        lines=10,
        placeholder="Enter your paragraph here...",
        label="Enter Text"
    ),

    outputs=[
        gr.Textbox(
            label="Analysis Result"
        ),

        gr.Plot(
            label="Word Frequency Graph"
        )
    ],

    title="Word Frequency Analyzer",

    description=(
        "Enter any paragraph and this NLP application "
        "will calculate word frequency and display the "
        "top 10 most frequent words."
    ),

    submit_btn="Analyze Text",
    clear_btn="Clear"
)


# STEP 6: Launch GUI
interface.launch()
```
